```python
import pandas as pd
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.common.keys import Keys
```

```python
# 正确的 reload 导入方式（Python 3.x）
from importlib import reload
import yuanbao  # 先导入模块
# 重新加载模块
yuanbao = reload(yuanbao)
# 从重新加载后的模块导入所有内容
from yuanbao import *
```

```python
def test(driver,text,num):
    yuanbao_send_p(driver,"请联网搜索："+text+"?"+'请以表格的形式输出。')
    print("--------------已发送问题 等待20s-----------------------")
    countdown_with_display(20)
    print("--------------点击窗口，开始收集数据-----------------------")
    check(driver,num)
    yuanbao_click_reference(driver)
    ActionChains(driver).send_keys(Keys.PAGE_DOWN).perform()
    countdown_with_display(1)
    time.sleep(1)
    DATA=collect_url(driver,text)
    print(DATA)
    countdown_with_display(1)
    print("--------------收集完毕，关闭窗口-----------------------")
    yuanbao_click_reference(driver)
    countdown_with_display(1)
    print('第一次对话完毕')
    return DATA
```

```python
#读取问句
df=pd.read_excel('问句.xlsx')
wenju_list=df['问句'].tolist()
# 启动浏览器
driver=init_chrome_driver_yuanbao()
```

```python
count=1
temp=0
res=pd.DataFrame()
wait = WebDriverWait(driver, 50)

for text in wenju_list:
    DATA=test(driver,text,count*2)
    count+=1
    if not DATA.empty:
        res = pd.concat([res, DATA], ignore_index=True)
    print(res.tail())
    time.sleep(1)
    if count%5==0:
        res.to_excel(f'yuanbao_data{temp}.xlsx',index=False)
        temp+=1

res.to_excel('yuanbao_data.xlsx',index=False)
print("数据已保存到 yuanbao_data.xlsx")
```
