```python
df=pd.read_excel('问句.xlsx')
wenju_list=df['问句'].tolist()
driver=init_chrome_driver_yuanbao()
```

```python
#测试单元
def test(text,num):
    doubao_send_p(driver,"请联网搜索："+text+"?"+"请用表格的形式输出。")
    print("--------------已发送问题 等待20s-----------------------")
    countdown_with_display(20)
    print("--------------点击窗口，开始收集数据-----------------------")
    check(driver,num)
    doubao_click_reference(driver)
    ActionChains(driver).send_keys(Keys.PAGE_DOWN).perform()
    countdown_with_display(1)
    DATA=doubao_collect_data(driver,text)
    print(DATA)
    countdown_with_display(1)
    print("--------------收集完毕，关闭窗口-----------------------")
    doubao_click_reference(driver)
    countdown_with_display(1)
    print('第一次对话完毕')
    return DATA
```

```python
count=1
temp=0
res=pd.DataFrame()
for text in wenju_list:
    DATA=test(text,count*2)
    count+=1
    if not DATA.empty:
        res = pd.concat([res, DATA], ignore_index=True)
    print(res.tail())
    time.sleep(1)
    if count%5==0:
        res.to_excel(f'doubao_data{temp}.xlsx',index=False)
        temp+=1
res.to_excel('doubao_data.xlsx',index=False)
print("数据已保存到 doubao_data.xlsx")
```
