```python
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.common.keys import Keys
```

```python
import deepseek
import importlib
importlib.reload(deepseek)
from deepseek import *
```

```python
driver=init_chrome_driver()
df=pd.read_excel('问句.xlsx')
wenju_list=df['问句'].tolist()
```

```python
#测试单元
def test(text,num):
    deepseek_send_p(driver,"请联网搜索："+text+"?"+'请用表格的形式输出。')
    print("--------------已发送问题 等待30s-----------------------")
    countdown_with_display(40)
    print("--------------点击窗口，开始收集数据-----------------------")
    check(driver,num)
    deepseek_click_reference(driver)
    ActionChains(driver).send_keys(Keys.PAGE_DOWN).perform()
    stop_deepseek(driver)
    countdown_with_display(1)
    DATA=deepseek_collect_data(driver,text)
    print(DATA)
    countdown_with_display(1)
    print("--------------收集完毕，关闭窗口-----------------------")
    deepseek_click_reference(driver)
    countdown_with_display(1)
    return DATA
```

```python
count=1
temp=0
res=pd.DataFrame()
wait = WebDriverWait(driver, 50)

for text in wenju_list:
    DATA=test(text,count*2)
    count+=1
    if not DATA.empty:
        res = pd.concat([res, DATA], ignore_index=True)
    print(res.tail())
    time.sleep(1)
    if count%5==0:
        res.to_excel(f'deepseek_data{temp}.xlsx',index=False)
        new_page_deepseek(driver)
        count=0
        time.sleep(2)

res.to_excel('deepseek_data.xlsx',index=False)

print("数据已保存到 deepseek_data.xlsx")
```
